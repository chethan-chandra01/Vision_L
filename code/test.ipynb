```python
import cv2
from ultralytics import YOLO

# 1. Load model and image
model = YOLO("../models/best.pt")
model.export(format="openvino")
image_path = "test_image.jpg"

# 2. Run inference on CPU
results = model.predict(source=image_path, device="cpu")

# 3. Generate the image with bounding rectangles drawn on it
annotated_frame = results[0].plot()

# 4. Save the output image and display it
cv2.imwrite("output_detected.jpg", annotated_frame)
cv2.imshow("VisionNav - Object Detection", annotated_frame)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
